# The Bias-Variance Tradeoff
####**Objective:** To understand and visualize the concepts of ***overfitting***, ***underfitting***, and the ***bias-variance tradeoff***,

    which are central to building models that generalize well to new data.
    In this lab, you will train models of varying complexity and plot their performance to see these concepts in action.

## Part 1:
**Understanding the Concepts**

  **Underfitting (High Bias):**  
             
      _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _
    | The model is too simple and fails to capture the underlying patterns   |
    | in the data. It performs poorly on both the training data and the test |
    | data. It has high bias because it makes strong assumptions, incorrect  |
    | about the data.                                                        |
    _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _  _ _ _ _ _ _ _ _ _ _ _ _ _                                               


The model over-indexes on specific data points (High Variance). It mistakes random noise for a permanent rule, making it 'brittle' when exposed to new, unseen data.

  **Overfitting (High Variance):**

      _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _
    | The model is **too complex** and learns the training data too well,          |
    | including the noise and random fluctuations. It performs exceptionally       |
    | well on the training data but poorly on the test data because it has         |
    | memorized the training set instead of learning the general pattern. It has   |
    | high variance because its performance changes drastically with different     |
    | training data.*                                                              |
     _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _
    
The model is structurally too rigid (High Bias) to recognize the data's trend. It essentially 'ignores' the evidence in favor of a pre-set, overly simple assumption.

---
###**The Goal:**

Find a model that is "just right"—complex enough to capture the true pattern but not so complex that it memorizes the noise. This is the **bias-variance tradeoff**.

The objective is 'Optimized Generalization.' We want a model that learns the 'S-curve' logic without becoming a slave to the individual outliers.

## Part 2:
**Setup**

To visualize this, we will create a synthetic (fake) dataset with a known pattern and some noise.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import learning_curve

# Generate some synthetic data
np.random.seed(0)
X = np.linspace(-5, 5, 100)
y = np.sin(X) + np.random.normal(0, 0.5, 100)
X = X[:, np.newaxis]

# Plot the data to see the underlying sine wave pattern
plt.figure(figsize=(10, 6))
plt.scatter(X, y, label='Data Points')
plt.title('Synthetic Sine Wave Data with Noise')
plt.legend()
plt.show()
```

# Part 3:
##Modeling with Different Complexities

We will use **Polynomial Regression** to control model complexity.
- A polynomial of degree 1 is a simple straight line (underfitting).
- A polynomial of degree 15 is a very complex, wiggly line (overfitting).

**Your Task:** \
Train and visualize three models with different degrees (1, 4, and 15) to see underfitting, a good fit, and overfitting.

# Task 1:
##Train and Plot Models

**Your Task:**

    For each degree (1, 4, 15), create a polynomial regression model, train it, and plot its predictions against the original data.

```python
# Now we train models of different complexity and compare them

plt.figure(figsize=(14, 8))
plt.scatter(X, y, label='Data Points')

degrees = [1, 4, 15]

for degree in degrees:
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    model.fit(X, y)

    # Use dynamic range instead of hardcoding
    X_plot = np.linspace(X.min(), X.max(), 1000)[:, np.newaxis]
    y_pred = model.predict(X_plot)

    plt.plot(X_plot, y_pred, label=f'Degree {degree}')

plt.legend()
plt.title('Comparing Model Complexity: Underfit vs Good Fit vs Overfit')
plt.xlabel('X')
plt.ylabel('y')
plt.ylim(-3, 3)
plt.show()
```

# Part 4:
#Learning Curves
**Concept:** \
A **learning curve** is a powerful tool to diagnose bias and variance. It plots the model's performance (e.g., accuracy or error) on both the **training set** and the **validation set** as a function of the number of training samples.*   
    
    High Bias (Underfitting):
    Both the training score and validation score will be low and will plateau quickly. The model is too simple to learn from more data.   
    
    High Variance (Overfitting):
    There will be a large gap between the high training score and the low validation score. The model memorized the training data but can't generalize.*   
    
    Just Right:
    The training and validation scores will converge to a high value.

## Task 2:
#Plot Learning Curves
**Your Task:** \
Use the `learning_curve` function from scikit-learn to plot the learning curves for the underfit (degree 1) and overfit (degree 15) models.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import ipywidgets as widgets
from IPython.display import display, Markdown

# Introductory text
display(Markdown("""Now that we’ve seen how different models fit the data, use the slider below to explore how model complexity changes the shape of the curve.

As you move the slider:
- Watch how the curve changes
- Look for when it starts matching the overall pattern
- Notice when it starts bending too much to hit individual points."""))

# 1. Function to plot polynomial fit and display guidance
def plot_poly(degree):
    # Create and train polynomial model
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    model.fit(X, y)

    # Generate smooth X for plotting
    X_plot = np.linspace(X.min(), X.max(), 1000)[:, np.newaxis]
    y_pred = model.predict(X_plot)

    # Dark-themed plot style
    plt.style.use('dark_background')
    plt.figure(figsize=(12, 6))

    # Plot data points with effect
    plt.scatter(X, y, color='#fffb00', edgecolors='black', s=60, label='Data Points', alpha=0.8)

    # Plot polynomial fit color
    plt.plot(X_plot, y_pred, color='#ff4f8b', linewidth=3, label=f'Polynomial Degree {degree}', alpha=0.9)

    # Customize the axes and grid for dark theme
    plt.title('Bias-Variance Tradeoff Visualization', color='#ffffff', fontsize=16)
    plt.xlabel('X', color='#ffffff')
    plt.ylabel('y', color='#ffffff')
    plt.ylim(-3, 3)
    plt.grid(color='#444444', linestyle='--', linewidth=0.5, alpha=0.7)
    plt.legend(facecolor='#222222', edgecolor='#ff4f8b', fontsize=12)
    plt.show()

    # Provide contextual instructions with themed style
   # Text explanation (clear, actionable, beginner-friendly)

    if degree <= 2:
        display(Markdown(f"""
### 🔵 Degree {degree}: Model is Too Simple

**What you see on the graph:**
- The line barely curves and ignores most of the data pattern.
- It misses the peaks and valleys completely.

**What this means:**
- The model is not flexible enough to learn the real relationship.
- It treats everything as roughly a straight trend.

**Why this is a problem:**
- Even on the training data, predictions are already wrong.
- It will also perform poorly on new data.

👉 **Takeaway:** If your model can't even match the general shape of your data, it is too simple.
"""))

    elif 3 <= degree <= 6:
        display(Markdown(f"""
### 🟢 Degree {degree}: Model is Learning the Right Pattern

**What you see on the graph:**
- The curve follows the overall shape of the data.
- It captures peaks and valleys but stays smooth.

**What this means:**
- The model is flexible enough to learn the real pattern.
- It ignores small random noise.

**Why this is good:**
- It fits the training data well without memorizing it.
- It is more likely to make accurate predictions on new data.

👉 **Takeaway:** A good model captures the pattern, not the noise.
"""))

    else:
        display(Markdown(f"""
### 🔴 Degree {degree}: Model is Overfitting

**What you see on the graph:**
- The curve twists sharply to pass very close to many individual points.
- You’ll notice lots of sudden ups and downs between nearby points.

**What this means:**
- The model is trying to fit *every single data point*, including random noise.
- It is no longer learning the true pattern — just memorizing the dataset.

**Why this is a problem:**
- It looks very accurate on this data…
- But small changes in data will cause completely different predictions.

👉 **Real-world consequence:** This model will likely fail when given new, unseen data.

👉 **Takeaway:** If your model bends sharply just to hit points exactly, it is memorizing, not learning.

"""))
# 2. Create interactive slider for polynomial degree
slider = widgets.IntSlider(value=1, min=1, max=15, step=1, description='Degree:')
_ = widgets.interact(plot_poly, degree=slider)
```
