**Data Ingestion & Conversion Script**\
The following code block handles the ingestion of the Zeek log and converts it into a standard, analysis-ready CSV format within the Colab environment.

```python
# =========================
#  IoT Malware EDA Pipeline
# Scenario 13 – Hakai / Mirai/Gafgyt
# =========================

# Install & Import Libraries
!pip install seaborn plotly --quiet

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

#  Load CSV
df = pd.read_csv('CTU-IoT-Malware-Capture-8-1.csv')

# -------------------------
#  Initial System Scan
# -------------------------
print(" Dataset Overview")
print("-" * 50)
print(f"Shape: {df.shape}")
print("\nColumn Info:")
print(df.info())
print("\nMissing Values:\n", df.isnull().sum())
print("\nFirst 5 Rows:\n", df.head())

# -------------------------
# 🛰 Data Cleaning
# -------------------------
numeric_cols = ['duration','orig_bytes','resp_bytes']

# Convert numeric columns
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Fill missing numeric values with median
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Convert timestamps & categorical features
df['ts'] = pd.to_datetime(df['ts'], errors='coerce')
df['proto'] = df['proto'].astype('category')
df['service'] = df['service'].astype('category')
df['label'] = df['label'].astype('category')

print("\n Data Cleaning Complete")

# -------------------------
#  Exploratory Data Analysis
# -------------------------

# Set style for "robotic/tech" theme
plt.style.use('dark_background')
sns.set_palette("coolwarm")
plt.rcParams['font.family'] = 'monospace'
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14

# 1️ Connection Volume Over Time
plt.figure(figsize=(14,5))
df.set_index('ts')['duration'].resample('H').count().plot(color='cyan')
plt.title("🛰 Connection Attempts per Hour")
plt.xlabel("Timestamp")
plt.ylabel("Number of Connections")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 2️ Bytes Distribution (Orig vs Resp)
plt.figure(figsize=(14,5))
sns.histplot(df['orig_bytes'], bins=50, color='lime', label='Orig Bytes', kde=True)
sns.histplot(df['resp_bytes'], bins=50, color='orange', label='Resp Bytes', kde=True)
plt.title("📡 Bytes Sent and Received Distribution")
plt.xlabel("Bytes")
plt.ylabel("Frequency")
plt.legend()
plt.show()

# 3️ Top 10 Communication Pairs
top_pairs = df.groupby(['id.orig_h','id.resp_h']).size().sort_values(ascending=False).head(10)
plt.figure(figsize=(12,6))
sns.barplot(x=top_pairs.values, y=[f"{o}->{r}" for o,r in top_pairs.index], palette="viridis")
plt.title("🔗 Top 10 Communicating IP Pairs")
plt.xlabel("Number of Connections")
plt.ylabel("IP Pairs")
plt.show()

# 4️ Protocol & Service Counts
fig, axes = plt.subplots(1,2, figsize=(14,5))
sns.countplot(data=df, x='proto', ax=axes[0], palette="plasma")
axes[0].set_title("⚡ Protocol Distribution")
sns.countplot(data=df, x='service', ax=axes[1], palette="magma")
axes[1].set_title("🛠 Service Distribution")
plt.tight_layout()
plt.show()

# 5️ Malware vs Benign Traffic
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='label', palette="cool")
plt.title(" Traffic Labels: Malware vs Benign")
plt.ylabel("Count")
plt.show()

# -------------------------
#  Outlier Detection & Feature Engineering
# -------------------------

# Log-transform bytes for heavy-tailed distributions
df['log_orig_bytes'] = np.log1p(df['orig_bytes'])
df['log_resp_bytes'] = np.log1p(df['resp_bytes'])

plt.figure(figsize=(14,5))
sns.boxplot(data=df[['log_orig_bytes','log_resp_bytes']], palette="inferno")
plt.title("🚨 Log-Transformed Byte Outliers")
plt.show()

# Example Feature Engineering
df['bytes_ratio'] = df['orig_bytes'] / (df['resp_bytes'] + 1)
df['duration_per_byte'] = df['duration'] / (df['orig_bytes'] + 1)

# -------------------------
#  Save Cleaned Dataset
# -------------------------
df.to_csv('CTU-IoT-Malware-Capture-8-1-clean.csv', index=False)
print("Cleaned dataset saved as 'CTU-IoT-Malware-Capture-8-1-clean.csv'")

```

```python
# =========================
# INTERACTIVE MALWARE DETECTION SCAN
# =========================

import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_csv('CTU-IoT-Malware-Capture-8-1-clean.csv')

# Remove emojis / non-ASCII characters
df.columns = df.columns.str.replace(r'[^\x00-\x7F]+','', regex=True)
str_cols = df.select_dtypes(include='object').columns
for col in str_cols:
    df[col] = df[col].str.replace(r'[^\x00-\x7F]+','', regex=True)

# Compute log-transformed bytes
df['log_orig_bytes'] = np.log1p(df['orig_bytes'])
df['log_resp_bytes'] = np.log1p(df['resp_bytes'])

# Compute ratio features
df['bytes_ratio'] = df['orig_bytes'] / (df['resp_bytes'] + 1)
df['duration_per_byte'] = df['duration'] / (df['orig_bytes'] + 1)

# =========================
#      Stepwise Scan
# =========================

insights = []

# 1. Network Persistence
high_freq_pairs = df.groupby(['id.orig_h','id.resp_h']).size().sort_values(ascending=False).head(5)
insights.append({
    'title': "NETWORK PERSISTENCE (Heartbeat Detection)",
    'description': "High-frequency IP pairs with failed connections may indicate C&C heartbeat.",
    'data': high_freq_pairs
})

# 2. Traffic Asymmetry
insights.append({
    'title': "TRAFFIC ASYMMETRY (Log-Scale Bytes)",
    'description': "Malware traffic is often one-sided. Log-transform of 'orig_bytes' and 'resp_bytes' helps normalize heavy-tailed data.",
    'data': f"log(orig_bytes) range [{df['log_orig_bytes'].min():.2f}, {df['log_orig_bytes'].max():.2f}]"
})

# 3. Protocol Anomalies
service_unknown = df[(df['service'] == '-') & (df['label'] == 'malware')]
insights.append({
    'title': "PROTOCOL ANOMALIES (Service vs Label)",
    'description': "Non-standard ports or unknown services often indicate C&C traffic.",
    'data': f"Packets with unknown service and malware label: {len(service_unknown)}"
})

# 4. Feature Engineering
insights.append({
    'title': "FEATURE ENGINEERING FOR ML",
    'description': "Ratio-based features expose behavioral signatures.",
    'data': f"bytes_ratio range [{df['bytes_ratio'].min():.2f}, {df['bytes_ratio'].max():.2f}]; duration_per_byte range [{df['duration_per_byte'].min():.2f}, {df['duration_per_byte'].max():.2f}]"
})

# =========================
# Step-through Display
# =========================

print("="*70)
print("STRATEGIC INSIGHTS FOR MALWARE DETECTION")
print("="*70)

for i, insight in enumerate(insights, start=1):
    print(f"\n{i}) {insight['title']}")
    print(f"   Description: {insight['description']}")
    print(f"   Data / Example: \n{insight['data']}")
    input("\nPress Enter to proceed to the next insight...")

print("="*70)
print("PIPELINE COMPLETE: Dataset is ready for Model Training.")
print("="*70)
```

```python
# -------------------------
# Data Ingestion & Conversion Script
# -------------------------
import pandas as pd

def convert_zeek_to_csv(input_file, output_file):
    """
    Parse Zeek/Bro logs by skipping metadata and producing a CSV ready for analysis
    """
    try:
        # Read as pipe-separated CSV
        df = pd.read_csv(input_file, sep='|', header=0, engine='python')
        df.to_csv(output_file, index=False)
        print(f"File saved as: {output_file}")
        return df
    except Exception as e:
        print(f"Error: {e}")
        return None

# Load dataset
df = convert_zeek_to_csv('CTU-IoT-Malware-Capture-8-1conn.log.labeled.csv',
                         'CTU-IoT-Malware-Capture-8-1.csv')
```
