Below we have loaded the data set saved as a  CSV file

```python
import pandas as pd
df = pd.read_csv("/content/CTU-IoT-Malware-Capture-8-1conn_Use_this_one-1 (1).csv")
display(df.head())
```

Below we replace invalid placeholder values with NaN so that missing or non-meaningful data can be properly identified and handled during preprocessing.

```python
import numpy as np
df.replace("-", np.nan, inplace=True)
```

We will remove identifiers that are not useful for Machine Learning:

```python
df.drop(columns=["uid","id.orig_h","id.resp_h","id.resp_p","history"],
errors="ignore", inplace=True)
```

And then convert numeric columns

```python
from sklearn.model_selection import train_test_split
import pandas as pd # Ensure pandas is imported

# Assuming 'df' is already defined from a previous cell.
# Split the main DataFrame into training and testing sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

cols_to_fix = ["duration","orig_bytes","orig_pkts"]

for col in cols_to_fix:
    train_df[col] = pd.to_numeric(train_df[col], errors="coerce")
    test_df[col] = pd.to_numeric(test_df[col], errors="coerce") # Also apply to test_df

train_medians = train_df[cols_to_fix].median()
```
