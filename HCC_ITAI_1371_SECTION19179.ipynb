```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/CTU-IoT-Malware-Capture-8-1conn_Use_this_one-1.csv')

# Drop problematic columns
df.drop(columns=["id.resp_p"], inplace=True, errors='ignore')
df.drop(columns=["history"], inplace=True, errors='ignore')

# Convert duration to numeric
df["duration"] = pd.to_numeric(df["duration"], errors="coerce")
df["duration"] = df["duration"].fillna(df["duration"].median())

# Fix service placeholder
df["service"] = df["service"].replace("-", "Unknown")

# Split dataset
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42
)

# Convert numeric columns
train_df["orig_bytes"] = pd.to_numeric(train_df["orig_bytes"], errors="coerce")
train_df["orig_pkts"] = pd.to_numeric(train_df["orig_pkts"], errors="coerce")

# Fill missing numeric values
numeric_cols = train_df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    train_df[col] = train_df[col].fillna(train_df[col].median())

# Feature engineering
train_df["bytes_per_packet"] = train_df["orig_bytes"] / (train_df["orig_pkts"] + 1)
train_df["packet_density"] = train_df["orig_pkts"] / (train_df["duration"] + 1)

# Drop high-cardinality identifiers
train_df.drop(columns=["uid","id.orig_h","id.resp_h"], inplace=True, errors="ignore")

# One-hot encoding
train_df = pd.get_dummies(train_df)

# Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_cols = train_df.select_dtypes(include=np.number).columns

train_df[numeric_cols] = scaler.fit_transform(train_df[numeric_cols])
```
