# Part 1: Regression vs. Classification
This is the most fundamental distinction between types of supervised learning problems.
**Regression:** The goal is to predict a **continuous numerical value**.      
   
*Examples:* Predicting the price of a house, the temperature tomorrow, or the stock price.
**Classification:** The goal is to predict a **discrete category or class label**.    
*Examples:* Predicting if an email is spam or not spam, if a flower is a setosa, versicolor, or virginica, or if a customer will churn or not.

**In this lab, we will tackle one of each.**

```python
import numpy as np
from sklearn.linear_model import LinearRegression, LogisticRegression

# --- STEP 1: PREPARE DATA ---
# Features: [Square Footage, Number of Bedrooms]
X = np.array([[1200, 2], [1500, 3], [2200, 3], [3500, 4]])

# --- STEP 2: REGRESSION (Continuous Value) ---
# Target: The actual price ($)
y_regression = [200000, 250000, 310000, 500000]

reg_model = LinearRegression().fit(X, y_regression)
house_prediction = reg_model.predict([[2800, 4]])

print(f"REGRESSION RESULT: Estimated House Value: ${house_prediction[0]:,.2f}")
# Output: A precise numerical value ($408,448.28)


# --- STEP 3: CLASSIFICATION (Categorical Label) ---
# Target: Is it a 'Luxury' home? (0 = No, 1 = Yes)
y_classification = [0, 0, 0, 1]

class_model = LogisticRegression().fit(X, y_classification)
luxury_prediction = class_model.predict([[2800, 4]])

status = "Luxury" if luxury_prediction[0] == 1 else "Standard"
print(f"CLASSIFICATION RESULT: This home is categorized as: {status}")
# Output: A specific group/label (Standard)
```

# Part 2: Linear Regression
**Concept:** Linear Regression is used to predict a continuous value. It works by finding the best-fitting straight line through the data points. The model learns a "slope" (coefficient) for each feature and an "intercept".

**Problem:** We will predict the `Fare` of a Titanic passenger based on their `Age` and `Pclass`.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Load and prepare the data
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
# For simplicity, we'll drop rows with missing age
df.dropna(subset=['Age'], inplace=True)

# Define features and target
features = ['Age', 'Pclass']
target_reg = 'Fare'
X_reg = df[features]
y_reg = df[target_reg]

# Split the data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# --- STEP 1: PREPARE FEATURES AND TARGET ---
# We use Age and Pclass (Passenger Class) to predict the Fare
# X (Features) = [[Age, Pclass]] | y (Target) = Fare ($)
data = {
    'Age': [22, 38, 26, 35, 54],
    'Pclass': [3, 1, 3, 1, 1],
    'Fare': [7.25, 71.28, 7.92, 53.10, 81.85]
}
df = pd.DataFrame(data)

X = df[['Age', 'Pclass']]
y = df['Fare']

# --- STEP 2: TRAIN THE MODEL ---
# The model finds the 'slope' for Age/Pclass and the 'intercept'
model = LinearRegression()
model.fit(X, y)

# --- STEP 3: PREDICT A CONTINUOUS VALUE ---
# Prediction for a 30-year-old in 2nd Class
new_passenger = [[30, 2]]
predicted_fare = model.predict(new_passenger)

print(f"Regression Result: Predicted Fare is ${predicted_fare[0]:.2f}")
# Output: A precise number, e.g., $32.45
```

**3.  Make predictions on the test data.4.  Evaluate the model using `mean_squared_error`. This metric tells us the average of the squared differences between the predicted and actual values.**

```python
import numpy as np

# 1. Create the model instance
lr_model = LinearRegression()

# 2. Train the model
lr_model.fit(X_train_reg, y_train_reg)

# 3. Make predictions
y_pred_reg = lr_model.predict(X_test_reg)

# 4. Evaluate the model
mse = mean_squared_error(y_test_reg, y_pred_reg)

print(f"Mean Squared Error for Fare Prediction: {mse:.2f}")
print(f"The square root of this is {np.sqrt(mse):.2f}, meaning our model is off by about ${np.sqrt(mse):.2f} on average.")
```

# Part 3: Logistic Regression

**Concept:** Despite its name, Logistic Regression is a **classification** algorithm. It works by calculating the probability that a given input belongs to a certain class. It's one of the most widely used and interpretable classification models.*   

**Problem:** We will predict whether a passenger `Survived` based on their `Age`, `Pclass`, and `Sex`.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import pandas as pd # Ensure pandas is imported for pd.read_csv
from sklearn.model_selection import train_test_split # Ensure train_test_split is imported

# Reload and prepare the data for classification
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
# For simplicity, we'll drop rows with missing age
df.dropna(subset=['Age'], inplace=True)

# Prepare data for classification
# We need to encode 'Sex' column
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Define features and target
features_cls = ['Age', 'Pclass', 'Sex']
target_cls = 'Survived'
X_cls = df[features_cls]
y_cls = df[target_cls]

# Split the data
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)
```

# Task 2:

**Train and Evaluate a Logistic Regression Model**

**Your Task:**
    
    Create an instance of the `LogisticRegression` model.  
    Train the model using the classification training data.
    Make predictions on the test data.
    Evaluate the model using `accuracy_score`.

```python
# 1. Create the model instance
log_model = LogisticRegression()

# 2. Train the model
log_model.fit(X_train_cls, y_train_cls)

# 3. Make predictions
y_pred_cls = log_model.predict(X_test_cls)

# 4. Evaluate the model
accuracy = accuracy_score(y_test_cls, y_pred_cls)

print(f"Accuracy for Survival Prediction: {accuracy:.2%}")
```

**I. In your own words, what is the key difference between a regression problem and a classification problem?**

## Regression vs. Classification

In the world of machine learning, the goal is often to teach a computer to make predictions based on data. Most of these tasks fall into two main categories: Regression and Classification.

Think of it this way:  
## Regression = "How much?"
## Classification = "What kind?"

---

     Regression: Predicting the Spectrum

    A regression problem is used when the output you want to predict is a continuous number. "Continuous" means the value can fall anywhere within a range and can include decimal values.

### Goal
To find relationships between variables in order to predict a specific quantity.

### Real-World Examples
- **Real Estate:** Predicting a house price (e.g., $450,250)  
- **Meteorology:** Predicting temperature (e.g., 22.5°C)  
- **Finance:** Estimating future stock prices  

---

     Classification: Sorting into Buckets

    A classification problem is used when the output is a category or label. Instead of predicting a number, you are predicting a type.

### Goal
To separate data into categories and determine which group a new data point belongs to.

### Real-World Examples
- **Healthcare:** "Malignant" vs. "Benign" tumors  
- **Email:** "Spam" vs. "Inbox"  
- **Image Recognition:** "Cat," "Dog," or "Bird"

**III. Why did we use mean_squared_error to evaluate the regression model but accuracy_score for the classification model? Why wouldn't accuracy be a good metric for the fare prediction task?**

## Evaluating Regression vs. Classification Models

Different types of problems require different evaluation metrics.

### Mean Squared Error (Regression)
We use mean_squared_error for regression because it measures how far off our predicted numerical values are from the actual values.

- It calculates the average squared difference between predicted and true values  
- Lower values indicate better model performance  

### Accuracy (Classification)
We use accuracy for classification because it measures how often the model correctly predicts categories.

- It calculates the percentage of correct predictions  
- Higher values indicate better performance  

### Why Not Use Accuracy for Regression?
Accuracy is not suitable for regression tasks like fare prediction because:

- The output is continuous (numerical)  
- Predictions are rarely exactly equal to the true value  

As a result, accuracy would almost always be very low and not meaningful for evaluating regression models.
